# Week 4 - Baseline Model

## 1. Import libraries

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

## 2. Load train and test set

```python
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")

y_train = pd.read_csv("../data/processed/y_train.csv")["ClosePrice"]
y_test = pd.read_csv("../data/processed/y_test.csv")["ClosePrice"]
```

## 3. Sanity Checks

Check matching row counts, identical feature order, and no missing values.

```python
print("Training shape:", X_train.shape)
print("Test shape:", X_test.shape)

assert len(X_train) == len(y_train), "Training inputs and targets have different lengths."
assert len(X_test) == len(y_test), "Test inputs and targets have different lengths."
assert X_train.columns.equals(X_test.columns), "Training and test columns do not match."
assert not X_train.isna().any().any(), "Training predictors contain missing values."
assert not X_test.isna().any().any(), "Test predictors contain missing values."
assert not y_train.isna().any(), "Training targets contain missing values."
assert not y_test.isna().any(), "Test targets contain missing values."

print("Data checks passed.")
```

## 4. Train the linear regression model

```python
baseline_model = LinearRegression()
baseline_model.fit(X_train, y_train)
```

## 5. Predict prices, evaluate test $R^2$

`.predict` returns an estimated sale price per test property.
 `r2_score` then compares our predictions against y_test. An $R^2$ score of 1 means our predictions were perfect. An $R^2$ score of 0 means our model ties with just guessing the average price for each property. 

```python
y_pred = baseline_model.predict(X_test)
test_r2 = r2_score(y_test, y_pred)

print(f"Linear Regression test R²: {test_r2:.4f}")
```

## 6. Record baseline result

Save the model name, score, and dataset sizes.

```python
baseline_results = pd.DataFrame({
    "Model": ["Linear Regression"],
    "Test_R2": [test_r2],
    "TrainingRows": [len(X_train)],
    "TestRows": [len(X_test)],
    "PredictorColumns": [X_train.shape[1]]
})

display(baseline_results)
baseline_results.to_csv("../data/processed/baseline_results.csv", index=False)
```

## 7. Summary

### Steps completed

- Loaded the training and test datasets prepared in Notebook 02.
- Checked that predictor columns matched, input and target lengths
  matched, and no missing values remained.
- Trained a Linear Regression model using 117,341 properties
  and 1,340 predictor columns.
- Predicted closing prices for 12,279 properties in the test set.
- Calculated test R² and saved the results to `baseline_results.csv`.
- Compared individual predicted prices with actual closing prices.

### Findings

The Linear Regression model achieved a test R² of **0.815914**.
This corresponds to approximately **81.6% less squared prediction
error** than predicting the test set's average closing price for
every property.
